### Problem 1 (9)

For each of the following systems models, find (i) any equilibria and
(ii) their stabilities.

#### Problem 1.1 (3)

$$\frac{dx}{dt} = 4x^2 - 16$$

### Answer to Problem 1.1
**(i)** To find the equilibria, let dx/dt = f(x) and find when f(x) = 0. Solving for $4x^2 = 16$, then I get x = +2, -2. So **the equilibria are x = 2 and x = -2.**

**(ii)** To determine the stability, find f'(x) and evaluate it at the two equilibria. $f'(x) = 8x$  Therefore, **the equilibrium at x = 2 is unstable because $f'(2) = 16 > 0$, while the equilibrium at x = -2 is stable because $f'(-2) = -16 < 0$.**

#### Problem 1.2 (3)

$$\frac{dx}{dt} = x - x^3$$


### Answer to Problem 1.2
**(i)** To find the equilibria, let dx/dt = f(x) and find when f(x) = 0. Solving for $x^3 = x$, I get x = -1, 0, 1. So **the equilibria are x = -1, x = 0 and x = 1.**

**(ii)** To determine the stability, find f'(x) and evaluate it at the two equilibria. $f'(x) =1-2(x^2)$ Therefore, **the equilibrium at x = -1 is stable because f'(-1) = -1 < 0, and the equilibrium at x = 0 is unstable because f'(0) = 1 > 0, while the equilibrium at x = 1 is stable because f'(1) = -1 < 0.**
